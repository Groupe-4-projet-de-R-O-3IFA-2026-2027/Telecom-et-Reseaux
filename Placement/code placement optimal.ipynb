```python
# -*- coding: utf-8 -*-
"""
PROJET : Placement de machines virtuelles (VM) sur des serveurs physiques.

Le programme :
  1. compare six méthodes (First Fit, Best Fit, First Fit Decreasing et trois modèles exacts)
     sur trois instances de test ;
  2. ouvre une session interactive : on saisit (ou on génère) ses VM UNE SEULE FOIS, puis on
     peut changer de méthode, comparer toutes les méthodes, ajouter ou supprimer une VM,
     sans jamais tout ressaisir.

Installation des dépendances :
    pip install "pulp<4" matplotlib       (PuLP 4 a changé son API)

Exécution en terminal :
    python placement_vm.py
    python placement_vm.py --sans-experiences     # va directement à la session interactive
    python placement_vm.py --sans-interactif      # n'exécute que les expériences

Dans Jupyter :
    %run placement_vm.py
"""

import argparse
import math
import random
import sys
import time

import matplotlib.pyplot as plt
import pulp
from matplotlib.patches import FancyBboxPatch

# Évite les erreurs d'affichage dans la console Windows.
if hasattr(sys.stdout, "reconfigure"):
    sys.stdout.reconfigure(encoding="utf-8")


# ============================================================================
# 1. PARAMETRES GENERAUX ET CATALOGUE DES SERVEURS
# ============================================================================

RESSOURCES = ("cpu", "ram", "disque")
ALPHA = 0.5  # poids du nombre de serveurs dans l'objectif pondéré
BETA = 0.5  # poids de la puissance dans l'objectif pondéré
TEMPS_MAX_EXACT = 60  # limite de temps (secondes) pour chaque modèle exact

# Catalogue de référence utilisé pour générer les inventaires synthétiques.
CATALOGUE_SERVEURS = [
    {
        "id": "S1",
        "cpu": 8,
        "ram": 32,
        "disque": 200,
        "p_repos": 100.0,
        "p_max": 250.0,
        "disponible": True,
    },
    {
        "id": "S2",
        "cpu": 16,
        "ram": 64,
        "disque": 500,
        "p_repos": 150.0,
        "p_max": 400.0,
        "disponible": True,
    },
    {
        "id": "S3",
        "cpu": 4,
        "ram": 16,
        "disque": 100,
        "p_repos": 60.0,
        "p_max": 150.0,
        "disponible": False,
    },
]


def attribuer_parametres_puissance(serveur):
    """Ajoute des paramètres de puissance simulés selon la capacité CPU."""
    s = serveur.copy()
    if s["cpu"] <= 8:
        s["p_repos"], s["p_max"] = 80.0, 160.0
    elif s["cpu"] <= 16:
        s["p_repos"], s["p_max"] = 120.0, 240.0
    else:
        s["p_repos"], s["p_max"] = 180.0, 360.0
    return s


def serveurs_disponibles(serveurs):
    """Sous-liste des serveurs disponibles."""
    return [s for s in serveurs if s.get("disponible", True)]


def valider_instance(vms, serveurs):
    """Vérifie identifiants uniques et valeurs strictement positives (lève ValueError)."""
    ids = set()
    for s in serveurs:
        if s["id"] in ids:
            raise ValueError(f"Identifiant de serveur dupliqué : {s['id']}")
        ids.add(s["id"])
        if s.get("disponible", True):
            for r in RESSOURCES:
                if r not in s or s[r] <= 0:
                    raise ValueError(f"Capacité invalide : {s['id']} - {r}")

    ids = set()
    for vm in vms:
        if vm["id"] in ids:
            raise ValueError(f"Identifiant de VM dupliqué : {vm['id']}")
        ids.add(vm["id"])
        for r in RESSOURCES:
            if r not in vm or vm[r] <= 0:
                raise ValueError(f"Besoin invalide : {vm['id']} - {r}")


# ============================================================================
# 2. GENERATION DES INSTANCES EXPERIMENTALES
# ============================================================================


def generer_instance_aleatoire(nb_serveurs=20, nb_vms=35, graine=42):
    """Génère une instance aléatoire reproductible (graine=None : vraiment aléatoire)."""
    hasard = random.Random(graine)
    modeles = [s for s in CATALOGUE_SERVEURS if s["disponible"]]

    serveurs_gen = []
    for i in range(1, nb_serveurs + 1):
        modele = hasard.choice(modeles)
        serveur = {
            "id": f"S{i}",
            "cpu": modele["cpu"],
            "ram": modele["ram"],
            "disque": modele["disque"],
            "disponible": hasard.random() < 0.8,
        }
        serveurs_gen.append(attribuer_parametres_puissance(serveur))

    vms_gen = [
        {
            "id": f"V{i}",
            "cpu": hasard.randint(1, 20),
            "ram": hasard.randint(1, 80),
            "disque": hasard.randint(10, 600),
        }
        for i in range(1, nb_vms + 1)
    ]
    return serveurs_gen, vms_gen


def generer_instance_realisable(serveurs_reference, nb_vms=35, graine=43):
    """
    Génère des VM individuellement compatibles avec au moins un serveur.
    Cela ne garantit pas que toutes les VM puissent être placées simultanément.
    """
    hasard = random.Random(graine)
    serveurs_gen = [s.copy() for s in serveurs_reference]
    disponibles = serveurs_disponibles(serveurs_gen)
    if not disponibles:
        raise ValueError("Aucun serveur disponible pour générer les VM.")

    vms_gen = []
    for i in range(1, nb_vms + 1):
        reference = hasard.choice(disponibles)
        vms_gen.append(
            {
                "id": f"R{i}",
                "cpu": hasard.randint(1, reference["cpu"]),
                "ram": hasard.randint(1, reference["ram"]),
                "disque": hasard.randint(1, reference["disque"]),
            }
        )
    return serveurs_gen, vms_gen


# ============================================================================
# 3. HEURISTIQUES : FIRST FIT, BEST FIT, FIRST FIT DECREASING
# ============================================================================


def _resultat_heuristique(placement, non_placees, restant):
    """Dictionnaire de résultat commun aux heuristiques."""
    utilises = list(dict.fromkeys(placement.values()))
    return {
        "statut": "realisable" if not non_placees else "placement_incomplet",
        "placement": placement,
        "vms_non_placees": non_placees,
        "capacites_restantes": restant,
        "serveurs_utilises": utilises,
        "nombre_serveurs_utilises": len(utilises),
    }


def first_fit(vms, serveurs):
    """Chaque VM va sur le premier serveur (dans l'ordre) qui peut l'accueillir."""
    valider_instance(vms, serveurs)
    disponibles = serveurs_disponibles(serveurs)
    restant = {s["id"]: {r: s[r] for r in RESSOURCES} for s in disponibles}

    placement, non_placees = {}, []
    for vm in vms:
        for s in disponibles:
            sid = s["id"]
            if all(vm[r] <= restant[sid][r] for r in RESSOURCES):
                for r in RESSOURCES:
                    restant[sid][r] -= vm[r]
                placement[vm["id"]] = sid
                break
        else:
            non_placees.append(vm["id"])

    return _resultat_heuristique(placement, non_placees, restant)


def best_fit(vms, serveurs):
    """Chaque VM va sur le serveur qui garde le moins de place libre après son arrivée."""
    valider_instance(vms, serveurs)
    disponibles = serveurs_disponibles(serveurs)
    restant = {s["id"]: {r: s[r] for r in RESSOURCES} for s in disponibles}
    capacites = {s["id"]: {r: s[r] for r in RESSOURCES} for s in disponibles}

    placement, non_placees = {}, []
    for vm in vms:
        meilleur, meilleur_score = None, None
        for s in disponibles:
            sid = s["id"]
            if not all(vm[r] <= restant[sid][r] for r in RESSOURCES):
                continue
            score = sum((restant[sid][r] - vm[r]) / capacites[sid][r] for r in RESSOURCES)
            if meilleur_score is None or score < meilleur_score:
                meilleur, meilleur_score = sid, score

        if meilleur is None:
            non_placees.append(vm["id"])
        else:
            for r in RESSOURCES:
                restant[meilleur][r] -= vm[r]
            placement[vm["id"]] = meilleur

    return _resultat_heuristique(placement, non_placees, restant)


def first_fit_decreasing(vms, serveurs, critere="score"):
    """First Fit après avoir trié les VM de la plus grosse à la plus petite."""
    criteres_valides = (*RESSOURCES, "score")
    if critere not in criteres_valides:
        raise ValueError(f"Critère invalide. Choisir parmi {criteres_valides}.")

    capacites_max = {
        r: max((s[r] for s in serveurs_disponibles(serveurs)), default=1) for r in RESSOURCES
    }

    def cle_tri(vm):
        if critere in RESSOURCES:
            return vm[critere]
        return sum(vm[r] / capacites_max[r] for r in RESSOURCES)

    return first_fit(sorted(vms, key=cle_tri, reverse=True), serveurs)


# ============================================================================
# 4. MODELES EXACTS AVEC PuLP / CBC
# ============================================================================


def objectif_minimiser_puissance(y, x, serveurs, vms):
    """Puissance totale : puissance au repos + part proportionnelle au CPU utilisé."""
    expression = 0
    for s in serveurs_disponibles(serveurs):
        sid = s["id"]
        if s.get("p_repos") is None or s.get("p_max") is None:
            raise ValueError(f"Paramètres p_repos/p_max manquants pour le serveur {sid}.")
        expression += s["p_repos"] * y[sid]
        expression += ((s["p_max"] - s["p_repos"]) / s["cpu"]) * pulp.lpSum(
            vm["cpu"] * x[(vm["id"], sid)] for vm in vms
        )
    return expression


def objectif_pondere(y, x, serveurs, vms):
    """Compromis ALPHA x (nombre de serveurs) + BETA x (puissance), chacun normalisé."""
    disponibles = serveurs_disponibles(serveurs)
    n_max = max(len(disponibles), 1)
    p_max_total = sum(s["p_max"] for s in disponibles)
    if p_max_total <= 0:
        raise ValueError("Puissance maximale totale invalide.")

    nombre_serveurs = pulp.lpSum(y.values())
    puissance = objectif_minimiser_puissance(y, x, serveurs, vms)
    return ALPHA * nombre_serveurs / n_max + BETA * puissance / p_max_total


def construire_modele_exact(serveurs, vms, objectif=None):
    """
    Modèle linéaire en variables binaires : x[vm, serveur] = 1 si la VM est sur ce serveur,
    y[serveur] = 1 si le serveur est allumé. Sans `objectif`, on minimise le nombre de serveurs.
    """
    valider_instance(vms, serveurs)
    disponibles = serveurs_disponibles(serveurs)
    ids_s = [s["id"] for s in disponibles]

    modele = pulp.LpProblem("placement_vm", pulp.LpMinimize)
    x = {
        (vm["id"], s["id"]): pulp.LpVariable(f"x_{i}_{j}", cat="Binary")
        for i, vm in enumerate(vms)
        for j, s in enumerate(disponibles)
    }
    y = {s["id"]: pulp.LpVariable(f"y_{j}", cat="Binary") for j, s in enumerate(disponibles)}

    if objectif is None:
        modele += pulp.lpSum(y.values()), "minimiser_nombre_serveurs"
    else:
        modele += objectif(y, x, disponibles, vms), "objectif"

    # Chaque VM est affectée à exactement un serveur.
    for vm in vms:
        modele += (
            pulp.lpSum(x[(vm["id"], sid)] for sid in ids_s) == 1,
            f"affectation_{vm['id']}",
        )

    # Capacités CPU, RAM et disque (nulles si le serveur est éteint).
    for s in disponibles:
        sid = s["id"]
        for r in RESSOURCES:
            modele += (
                pulp.lpSum(vm[r] * x[(vm["id"], sid)] for vm in vms) <= s[r] * y[sid],
                f"capacite_{r}_{sid}",
            )

    # Une affectation implique l'activation du serveur.
    for (vid, sid), variable in x.items():
        modele += variable <= y[sid], f"activation_{vid}_{sid}"

    return modele, x, y


def resoudre_modele_exact(modele, temps_max=TEMPS_MAX_EXACT, afficher_log=False):
    """Résout avec CBC. Renvoie « optimal », « faisable », « infaisable » ou « erreur »."""
    solveur = pulp.PULP_CBC_CMD(msg=afficher_log, timeLimit=temps_max)
    try:
        modele.solve(solveur)
    except Exception as erreur:
        print("Erreur du solveur CBC :", erreur)
        return "erreur"

    if modele.status == pulp.LpStatusOptimal and modele.sol_status == pulp.LpSolutionOptimal:
        return "optimal"
    if modele.sol_status == pulp.LpSolutionIntegerFeasible:
        return "faisable"
    if modele.status == pulp.LpStatusInfeasible:
        return "infaisable"
    return "erreur"


def extraire_placement_exact(x, vms, serveurs):
    """Traduit les variables x en dictionnaire {id de VM: id de serveur}."""
    placement = {}
    for vm in vms:
        for s in serveurs:
            variable = x.get((vm["id"], s["id"]))
            if variable is not None and (variable.value() or 0) > 0.5:
                placement[vm["id"]] = s["id"]
                break
    return placement


# ============================================================================
# 5. CALCUL DE PUISSANCE ET AFFICHAGE DES RESULTATS
# ============================================================================


def calculer_puissance_placement(placement, serveurs, vms):
    """Puissance totale (W) des serveurs allumés : repos + part proportionnelle au CPU utilisé."""
    puissance_totale = 0.0
    utilises = set(placement.values())

    for s in serveurs:
        sid = s["id"]
        if sid not in utilises:
            continue
        if s.get("p_repos") is None or s.get("p_max") is None:
            raise ValueError(f"Puissance inconnue pour {sid}.")
        cpu_utilise = sum(vm["cpu"] for vm in vms if placement.get(vm["id"]) == sid)
        puissance_totale += s["p_repos"] + (s["p_max"] - s["p_repos"]) * cpu_utilise / s["cpu"]
    return puissance_totale


def puissance_du_resultat(resultat, vms, serveurs):
    """Puissance du placement, ou None si le placement est incomplet ou la puissance inconnue."""
    placement = resultat["placement"]
    if not placement or len(placement) < len(vms):
        return None
    try:
        return calculer_puissance_placement(placement, serveurs, vms)
    except ValueError:
        return None


def afficher_resultats_utilisateur(resultat, vms, serveurs):
    """Affiche le placement VM par VM, le bilan de chaque serveur et les indicateurs globaux."""
    placement = resultat.get("placement", {})

    print("\n========== RESULTATS DU PLACEMENT ==========")
    for vm in vms:
        sid = placement.get(vm["id"])
        print(f"{vm['id']} --> {sid if sid else 'NON PLACEE'}")

    if not placement:
        print("Aucune VM n'a été placée.")
        return

    print("\n========== BILAN DES SERVEURS ==========")
    nb_serveurs = 0
    for s in serveurs:
        vms_s = [vm for vm in vms if placement.get(vm["id"]) == s["id"]]
        if not vms_s:
            continue
        nb_serveurs += 1
        print(f"\nServeur : {s['id']}")
        print("VM accueillies :", [vm["id"] for vm in vms_s])
        for r, nom in (("cpu", "CPU"), ("ram", "RAM"), ("disque", "Disque")):
            utilise = sum(vm[r] for vm in vms_s)
            print(f"{nom} : {utilise}/{s[r]} (reste {s[r] - utilise})")

    print("\n========== INDICATEURS GLOBAUX ==========")
    print("VM placées :", len(placement), "/", len(vms))
    print("Serveurs utilisés :", nb_serveurs)
    puissance = puissance_du_resultat(resultat, vms, serveurs)
    if puissance is not None:
        print("Puissance estimée :", round(puissance, 2), "W")
    elif len(placement) == len(vms):
        print("Puissance non calculée (paramètres de puissance manquants).")


def visualiser_placement(resultat, vms, serveurs):
    """Dessine chaque serveur utilisé avec les VM qu'il accueille."""
    placement = resultat.get("placement", {})
    serveurs_utilises = [s for s in serveurs if s["id"] in placement.values()]
    if not serveurs_utilises:
        print("Aucun placement à représenter.")
        return

    fig, ax = plt.subplots(figsize=(12, max(4, 2.3 * len(serveurs_utilises))))
    y = 0

    for s in serveurs_utilises:
        vms_s = [vm for vm in vms if placement.get(vm["id"]) == s["id"]]
        hauteur = 1.2 + math.ceil(len(vms_s) / 2) * 0.9

        ax.add_patch(
            FancyBboxPatch(
                (0.2, y),
                8,
                hauteur,
                boxstyle="round,pad=0.08",
                facecolor="#E8EEF5",
                edgecolor="#34495E",
                linewidth=2,
            )
        )
        ax.text(
            4.2,
            y + 0.35,
            f"SERVEUR {s['id']} | CPU {s['cpu']} | RAM {s['ram']} Go | DISQUE {s['disque']} Go",
            ha="center",
            fontsize=9,
            fontweight="bold",
        )

        for i, vm in enumerate(vms_s):
            x_vm = 0.5 + (i % 2) * 3.9
            y_vm = y + 0.65 + (i // 2) * 0.9
            ax.add_patch(
                FancyBboxPatch(
                    (x_vm, y_vm),
                    3.5,
                    0.7,
                    boxstyle="round,pad=0.04",
                    facecolor="#B8D8F0",
                    edgecolor="#2471A3",
                    linewidth=1.3,
                )
            )
            ax.text(
                x_vm + 1.75,
                y_vm + 0.35,
                f"{vm['id']} | CPU:{vm['cpu']} RAM:{vm['ram']} DISQUE:{vm['disque']}",
                ha="center",
                va="center",
                fontsize=8,
            )
        y += hauteur + 0.3

    ax.set_xlim(0, 8.5)
    ax.set_ylim(y + 0.2, -0.2)
    ax.set_title("Représentation du placement des machines virtuelles")
    ax.axis("off")
    plt.tight_layout()
    plt.show()


# ============================================================================
# 6. LES SIX METHODES : EXECUTION UNIFORME ET COMPARAISON
# ============================================================================


def _methode_exacte(objectif):
    """Fabrique la fonction (vms, serveurs) -> résultat d'un modèle exact."""

    def lancer(vms, serveurs):
        modele, x, _ = construire_modele_exact(serveurs, vms, objectif)
        statut = resoudre_modele_exact(modele)
        placement = (
            extraire_placement_exact(x, vms, serveurs) if statut in ("optimal", "faisable") else {}
        )
        return {"statut": statut, "placement": placement}

    return lancer


# Registre des méthodes : pour en ajouter une, il suffit d'ajouter une ligne ici.
METHODES = {
    "First Fit": first_fit,
    "Best Fit": best_fit,
    "First Fit Decreasing": first_fit_decreasing,
    "Exact - Nombre de serveurs": _methode_exacte(None),
    "Exact - Puissance": _methode_exacte(objectif_minimiser_puissance),
    "Exact - Pondéré": _methode_exacte(objectif_pondere),
}
NOMS_METHODES = list(METHODES)


def executer_une_methode(nom, vms, serveurs):
    """Exécute la méthode `nom` sur une copie des données ; ajoute le temps (et l'erreur éventuelle)."""
    serveurs_test = [s.copy() for s in serveurs]
    debut = time.perf_counter()
    try:
        resultat = METHODES[nom](vms, serveurs_test)
    except Exception as erreur:
        resultat = {"statut": "erreur", "placement": {}, "erreur": str(erreur)}
    resultat["temps"] = time.perf_counter() - debut
    return resultat


def afficher_tableau_comparatif(resultats, vms, serveurs):
    """Tableau récapitulatif : une ligne par méthode."""
    largeur = 104
    print("\n" + "-" * largeur)
    print(
        f"{'Méthode':28} {'Statut':>20} {'VM placées':>11} {'Serveurs':>9} "
        f"{'Puissance (W)':>14} {'Temps (s)':>10}"
    )
    print("-" * largeur)
    for nom, resultat in resultats.items():
        placement = resultat["placement"]
        puissance = puissance_du_resultat(resultat, vms, serveurs)
        texte_puissance = f"{puissance:.2f}" if puissance is not None else "-"
        print(
            f"{nom:28} {resultat['statut']:>20} {len(placement):>11} "
            f"{len(set(placement.values())):>9} {texte_puissance:>14} {resultat['temps']:>10.4f}"
        )
    print("-" * largeur)


def comparer_toutes_les_methodes(vms, serveurs, detail=False):
    """Exécute les six méthodes sur les mêmes VM et affiche le tableau comparatif."""
    resultats = {}
    for nom in NOMS_METHODES:
        resultat = executer_une_methode(nom, vms, serveurs)
        resultats[nom] = resultat

        if detail:
            placement = resultat["placement"]
            print("\n" + "-" * 65)
            print("METHODE :", nom)
            print("Statut :", resultat["statut"])
            print(f"VM placées : {len(placement)}/{len(vms)}")
            print("Serveurs utilisés :", len(set(placement.values())))
            print(f"Temps d'exécution : {resultat['temps']:.6f} seconde(s)")
            if "erreur" in resultat:
                print("Erreur :", resultat["erreur"])
            if len(placement) < len(vms):
                print("VM non placées :", [vm["id"] for vm in vms if vm["id"] not in placement])
            puissance = puissance_du_resultat(resultat, vms, serveurs)
            if puissance is not None:
                print(f"Puissance estimée : {puissance:.2f} W")
            else:
                print("Puissance : non calculée (placement incomplet).")

    afficher_tableau_comparatif(resultats, vms, serveurs)
    return resultats


def afficher_comparaison_experience(titre, vms, serveurs):
    """Une expérience : en-tête, détail de chaque méthode, tableau comparatif."""
    print("\n" + "=" * 85)
    print(titre)
    print("=" * 85)
    print("Nombre de VM :", len(vms))
    print("Nombre de serveurs :", len(serveurs))
    print("Serveurs disponibles :", len(serveurs_disponibles(serveurs)))
    return comparer_toutes_les_methodes(vms, serveurs, detail=True)


def lancer_experiences():
    """Les trois expériences de comparaison."""
    print("\nPARTIE I : EXPERIMENTATIONS")

    serveurs_petits = [
        {
            "id": "S1",
            "cpu": 8,
            "ram": 16,
            "disque": 100,
            "disponible": True,
            "p_repos": 80.0,
            "p_max": 160.0,
        },
        {
            "id": "S2",
            "cpu": 4,
            "ram": 8,
            "disque": 50,
            "disponible": True,
            "p_repos": 50.0,
            "p_max": 100.0,
        },
        {
            "id": "S3",
            "cpu": 64,
            "ram": 256,
            "disque": 2000,
            "disponible": False,
            "p_repos": 100.0,
            "p_max": 200.0,
        },
    ]
    vms_petites = [
        {"id": "VM1", "cpu": 2, "ram": 4, "disque": 20},
        {"id": "VM2", "cpu": 2, "ram": 4, "disque": 30},
        {"id": "VM3", "cpu": 6, "ram": 12, "disque": 60},
    ]
    resultats_petits = afficher_comparaison_experience(
        "EXPERIENCE 1 : PETITE INSTANCE MANUELLE (3 VM, 3 SERVEURS)", vms_petites, serveurs_petits
    )

    serveurs_grands, vms_grandes = generer_instance_aleatoire(nb_serveurs=20, nb_vms=35, graine=42)
    resultats_grands = afficher_comparaison_experience(
        "EXPERIENCE 2 : INSTANCE ALEATOIRE DE 35 VM", vms_grandes, serveurs_grands
    )

    serveurs_realistes, vms_realistes = generer_instance_realisable(
        serveurs_grands, nb_vms=10, graine=43
    )
    resultats_10 = afficher_comparaison_experience(
        "EXPERIENCE 3 : INSTANCE REALISABLE DE 10 VM", vms_realistes, serveurs_realistes
    )

    return {
        "experience_petite": resultats_petits,
        "experience_grande": resultats_grands,
        "experience_10_VM": resultats_10,
    }


# ============================================================================
# 7. SESSION INTERACTIVE : SAISIE DES VM UNE FOIS, METHODE CHANGEABLE A VOLONTE
# ============================================================================


def preparer_base_serveurs():
    """Inventaire des serveurs de l'application (avec paramètres de puissance)."""
    base, _ = generer_instance_aleatoire(nb_serveurs=20, nb_vms=0, graine=42)
    disponibles = serveurs_disponibles(base)
    capacites_max = {r: max(s[r] for s in disponibles) for r in RESSOURCES}

    print("\n===== INVENTAIRE DES SERVEURS =====")
    print("Serveurs au total :", len(base))
    print("Serveurs disponibles :", len(disponibles))
    print("Serveurs indisponibles :", len(base) - len(disponibles))
    print("\nCapacités maximales individuelles :")
    print("CPU :", capacites_max["cpu"], "coeurs")
    print("RAM :", capacites_max["ram"], "Go")
    print("Disque :", capacites_max["disque"], "Go")
    return base, capacites_max


def saisir_entier(invite, minimum=1, maximum=None):
    """Demande un entier (entre minimum et maximum) jusqu'à obtenir une valeur valide."""
    while True:
        try:
            valeur = int(input(invite))
        except ValueError:
            print("Entrez un nombre entier.")
            continue
        if valeur < minimum:
            print(f"La valeur doit être au moins {minimum}.")
        elif maximum is not None and valeur > maximum:
            print(f"Le maximum disponible est {maximum}.")
        else:
            return valeur


def saisir_une_vm(vms_existantes, capacites_max, numero):
    """Saisit une VM (identifiant unique + CPU, RAM, disque dans les limites de l'inventaire)."""
    print(f"\n--- MACHINE VIRTUELLE {numero} ---")
    while True:
        identifiant = input("Identifiant : ").strip()
        if not identifiant:
            print("L'identifiant ne peut pas être vide.")
        elif any(vm["id"] == identifiant for vm in vms_existantes):
            print("Cet identifiant existe déjà.")
        else:
            break

    vm = {"id": identifiant}
    for r in RESSOURCES:
        vm[r] = saisir_entier(f"{r.upper()} requis : ", 1, capacites_max[r])
    return vm


def saisir_vms(capacites_max):
    """Saisit un nombre choisi de VM."""
    print("\n===== SAISIE DES MACHINES VIRTUELLES =====")
    nombre = saisir_entier("Combien de VM voulez-vous placer ? ")
    vms = []
    for i in range(nombre):
        vms.append(saisir_une_vm(vms, capacites_max, i + 1))
    return vms


def verifier_vms(vms, serveurs):
    """Vérifie que les VM sont valides et que chacune tient seule sur au moins un serveur."""
    if not vms:
        print("Aucune VM saisie.")
        return False

    ids = set()
    disponibles = serveurs_disponibles(serveurs)
    for vm in vms:
        if not vm["id"] or vm["id"] in ids:
            print("Identifiant vide ou répété :", vm["id"])
            return False
        ids.add(vm["id"])

        for r in RESSOURCES:
            if not isinstance(vm[r], int) or vm[r] <= 0:
                print("Besoin invalide :", vm["id"], r)
                return False

        if not any(all(vm[r] <= s[r] for r in RESSOURCES) for s in disponibles):
            print("Aucun serveur ne peut accueillir individuellement", vm["id"])
            return False
    return True


def afficher_vms(vms):
    """Liste les VM courantes."""
    print(f"\nVM courantes ({len(vms)}) :")
    for vm in vms:
        print(
            f"  {vm['id']:>8} : CPU {vm['cpu']:>3} | RAM {vm['ram']:>3} | Disque {vm['disque']:>4}"
        )


def obtenir_vms(serveurs, capacites_max):
    """Obtient une liste de VM valide : saisie manuelle ou génération aléatoire compatible."""
    while True:
        print("\nComment obtenir les VM ?")
        print("1. Les saisir à la main")
        print("2. Générer des VM aléatoires (compatibles avec l'inventaire)")
        choix = input("Votre choix (1 ou 2) : ").strip()

        if choix == "1":
            vms = saisir_vms(capacites_max)
        elif choix == "2":
            nombre = saisir_entier("Combien de VM générer ? ")
            _, vms = generer_instance_realisable(serveurs, nb_vms=nombre, graine=None)
        else:
            print("Choix invalide.")
            continue

        if verifier_vms(vms, serveurs):
            afficher_vms(vms)
            return vms
        print("La saisie n'est pas compatible avec l'inventaire. On recommence.")


def lancer_et_afficher(nom, vms, serveurs, avec_schema):
    """Exécute une méthode sur les VM courantes et affiche tout le résultat."""
    print(f"\n>>> Exécution : {nom}")
    resultat = executer_une_methode(nom, vms, serveurs)

    print("Méthode :", nom)
    print("Statut :", resultat["statut"])
    print(f"Temps : {resultat['temps']:.6f} seconde(s)")
    if "erreur" in resultat:
        print("Erreur :", resultat["erreur"])

    afficher_resultats_utilisateur(resultat, vms, serveurs)
    if avec_schema:
        visualiser_placement(resultat, vms, serveurs)
    return resultat


def afficher_menu(vms, avec_schema):
    print("\n" + "=" * 60)
    print(f"MENU  ({len(vms)} VM en mémoire, schéma {'activé' if avec_schema else 'désactivé'})")
    print("=" * 60)
    for i, nom in enumerate(NOMS_METHODES, start=1):
        print(f" {i}. Placer avec : {nom}")
    print(f" {len(NOMS_METHODES) + 1}. Comparer les {len(NOMS_METHODES)} méthodes sur ces VM")
    print(" A. Ajouter une VM        S. Supprimer une VM")
    print(" L. Lister les VM         N. Nouvelles VM (tout remplacer)")
    print(" G. Activer/désactiver le schéma        Q. Quitter")


def session_interactive(serveurs, capacites_max):
    """Les VM sont saisies une seule fois ; on enchaîne ensuite les méthodes à volonté."""
    vms = obtenir_vms(serveurs, capacites_max)
    avec_schema = True
    numero_comparaison = len(NOMS_METHODES) + 1

    while True:
        afficher_menu(vms, avec_schema)
        commande = input("Votre choix : ").strip().upper()

        if commande.isdigit() and 1 <= int(commande) <= len(NOMS_METHODES):
            lancer_et_afficher(NOMS_METHODES[int(commande) - 1], vms, serveurs, avec_schema)

        elif commande == str(numero_comparaison):
            print("\nComparaison en cours (les modèles exacts peuvent prendre du temps)...")
            comparer_toutes_les_methodes(vms, serveurs)

        elif commande == "A":
            nouvelle = saisir_une_vm(vms, capacites_max, len(vms) + 1)
            if verifier_vms(vms + [nouvelle], serveurs):
                vms.append(nouvelle)
                print(f"VM {nouvelle['id']} ajoutée.")
            else:
                print("VM refusée : elle n'a pas été ajoutée.")

        elif commande == "S":
            identifiant = input("Identifiant de la VM à supprimer : ").strip()
            restantes = [vm for vm in vms if vm["id"] != identifiant]
            if len(restantes) == len(vms):
                print("Identifiant introuvable.")
            elif not restantes:
                print("Impossible de supprimer la dernière VM : utilisez N pour recommencer.")
            else:
                vms = restantes
                print(f"VM {identifiant} supprimée.")

        elif commande == "L":
            afficher_vms(vms)

        elif commande == "N":
            vms = obtenir_vms(serveurs, capacites_max)

        elif commande == "G":
            avec_schema = not avec_schema
            print("Schéma", "activé." if avec_schema else "désactivé.")

        elif commande == "Q":
            break

        else:
            print("Choix invalide.")

    print("\nFin de la session.")


# ============================================================================
# 8. PROGRAMME PRINCIPAL
# ============================================================================


def main():
    parser = argparse.ArgumentParser(description="Placement de VM sur des serveurs physiques.")
    parser.add_argument("--sans-experiences", action="store_true", help="saute les expériences")
    parser.add_argument(
        "--sans-interactif", action="store_true", help="saute la session interactive"
    )
    args, _ = parser.parse_known_args()

    print("\n" + "#" * 85)
    print("PROJET : PLACEMENT DE MACHINES VIRTUELLES SUR DES SERVEURS PHYSIQUES")
    print("#" * 85)

    resultats = {}
    if not args.sans_experiences:
        resultats = lancer_experiences()

    if not args.sans_interactif:
        print("\nPARTIE II : APPLICATION INTERACTIVE")
        serveurs, capacites_max = preparer_base_serveurs()
        session_interactive(serveurs, capacites_max)

    return resultats


if __name__ == "__main__":
    main()
```
